```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
engine = create_engine(
    "postgresql://tethys_super:pass@localhost:5436/hydro_correlation_tool_primary_db"
)
```

```python
sql = 'SELECT usgs_id, gage_name, latitude, longitude, nwm_feature_id, seeded_nwm_feature_id, nwm_kge_rating, nwm_kge_shared_dates, geoglows_river_id, seeded_geoglows_river_id, geoglows_kge_rating, geoglows_kge_shared_dates, verification_status, last_modified_by, last_modified_timestamp FROM gage_mapping ORDER BY usgs_id'
```

```python
with engine.connect() as conn:
    test = conn.execute(sql)
    gage_mapping_df = pd.DataFrame(test.fetchall(), columns=test.keys())
```

```python
gage_mapping_df = gage_mapping_df.astype({'nwm_feature_id': 'Int64', 'seeded_nwm_feature_id': 'Int64', 'geoglows_river_id': 'Int64', 'seeded_geoglows_river_id': 'Int64', 'nwm_kge_shared_dates': 'Int64', 'geoglows_kge_shared_dates': 'Int64'})
```

```python
gage_mapping_df.to_csv('../tethysapp/hydro_correlation_tool/public/data/gage_mapping.csv', index=False)
```
