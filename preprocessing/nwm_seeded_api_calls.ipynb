```python
network = 'NWM'
records = []
start_date = '2018-01-01'
end_date = '2022-12-31'
api_key = open("nwm_token.txt").read().strip()
Session = sessionmaker(bind=engine)
session = Session() 
try:
    for id in series_500:
        row = session.get(cacheTable, (network, id))
        if not row:
            time.sleep(.5)
            start_time = time.perf_counter()
            data = get_nwm_retrospective(river_id=id,start=start_date,end=end_date,api_key=api_key)
            end_time = time.perf_counter()
            seconds = end_time - start_time
            if "error" in data:
                outcome = "error"
                n_rows = None
            elif not len(data['dates']):
                outcome = "no dates"
                n_rows = 0
            else:
                outcome = 'success'
                n_rows = len(data['dates'])
                new_row = cacheTable(network=network, reach_id=id, reach_data=data, start_date=start_date, end_date=end_date)
                session.add(new_row)
                session.commit()
        else:
            outcome = 'already cached'
            seconds = None
            n_rows = None
        records.append({'id': id, 'seconds': seconds, 'outcome': outcome, 'n_rows': n_rows})
        print(len(records), id, outcome, round(seconds or 0, 1))
                
finally:
    session.close()
    final_df = pd.DataFrame(records)
    final_df.to_csv('api_test_results.csv', index=False)

    
```
