```python
def batch_nwm_daily_series(river_ids, start, end, api_key):
    batch_size = river_ids.count(",") + 1
    nwm_params = {
        'reach_id': river_ids,
        'start_time': start,
        'end_time': end,
        'output_format': 'csv'}
    start_time = time.perf_counter()
    r=requests.get("https://nwm-api-v2-9f6idmxh.uc.gateway.dev/retrospectives",
                       params=nwm_params,
                       headers={'x-api-key': api_key},
                       timeout=1000)
    end_time = time.perf_counter()
    request_time = end_time - start_time
    start_time = time.perf_counter()
    r.raise_for_status()
    df = pd.read_csv(io.StringIO(r.text))
    reach_dict = {}
    df["time"] = pd.to_datetime(df["time"])
    for rid, g in df.groupby("reach_id"):
        daily = g.set_index("time")["streamflow"].resample("D").mean().dropna()
        dates = daily.index.strftime("%Y-%m-%d")
        reach_dict[rid] = {"dates": list(dates), "values": list(daily), "units": "m^3/s"}
    n_returned = len(reach_dict)
    end_time = time.perf_counter()
    parse_time = end_time - start_time
    final_df = batch_cache(reach_dict, request_time, parse_time, batch_size, n_returned)
    return (reach_dict, request_time, parse_time, batch_size, n_returned, final_df)

```

```python
def batch_cache(reach_dict, request_time, parse_time, batch_size, n_returned):
    request_seconds_per_reach = request_time / batch_size
    parse_seconds_per_reach = parse_time / batch_size
    network = 'NWM'
    records = []
    start_date = '2018-01-01'
    end_date = '2022-12-31'
    Session = sessionmaker(bind=engine)
    session = Session() 
    try:
        for key, value in reach_dict.items():
            row = session.get(cacheTable, (network, int(key)))
            if not row:
                if "error" in value:
                    outcome = "error"
                    n_rows = None
                elif not len(value['dates']):
                    outcome = "no dates"
                    n_rows = 0
                else:
                    outcome = 'success'
                    n_rows = len(value['dates'])
                    new_row = cacheTable(network=network, reach_id=key, reach_data=value, start_date=start_date, end_date=end_date)
                    session.add(new_row)
                    session.commit()
            else:
                outcome = 'already cached'
                n_rows = None
            records.append({'id': key, "request_time": request_time, "parse_time": parse_time, 'outcome': outcome, 'n_rows': n_rows, "batch_size": batch_size, "n_returned": n_returned, "request_seconds_per_reach": request_seconds_per_reach, "parse_seconds_per_reach": parse_seconds_per_reach})
            print(len(records), key, outcome, round(request_time or 0, 1))
                
    finally:
        session.close()
    return pd.DataFrame(records)
```

```python
all_records = []
for index, batch in enumerate(batches):
    try:
        reach_dict, request_time, parse_time, batch_size, n_returned, final_df = \
            batch_nwm_daily_series(river_ids=batch, start=start_date, end=end_date, api_key=api_key)
        all_records.append(final_df)
    except Exception as e:
        all_records.append(pd.DataFrame([{"batch_size": (batch.count(",") + 1), "outcome": "error", "message": str(e)}]))
        print(f"Failure on batch {index + 1}")
    pd.concat(all_records, ignore_index=True).to_csv('api_test_results_batches.csv', index=False)
    time.sleep(0.5)
```
