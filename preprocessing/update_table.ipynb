```python
import pandas as pd
import sqlalchemy as sa
from sqlalchemy import create_engine
```

```python
engine = create_engine("postgresql://tethys_super:pass@localhost:5436/hydro_correlation_tool_primary_db")
```

```python
seed = pd.read_csv('seed.csv')
```

```python
with engine.begin() as conn:
    for index, row in seed.iterrows():
        if (pd.isna(row['nwm_feature_id'])):
            nwm_row = None
        else:
            nwm_row = int(row['nwm_feature_id'])
        if (pd.isna(row['geoglows_river_id'])):
            geo_row = None
        else:
            geo_row = int(row['geoglows_river_id'])
        sql = sa.text('UPDATE gage_mapping SET seeded_nwm_feature_id = :nwm, seeded_geoglows_river_id = :geo WHERE usgs_id = :usgs_id')
        conn.execute(sql, {'nwm': nwm_row, 'geo': geo_row, 'usgs_id': row['usgs_id']})
```
